```python
# =================================================================
# 0.  GLOBAL SETTINGS  (keep these lines)
# =================================================================
import os, glob, random, numpy as np, librosa, torch
from tqdm import tqdm
SAMPLERATE = 8000; DURATION = 2.0; MAX_LEN = int(DURATION*SAMPLERATE); N_MELS = 64

def load_wave(path):
    y, sr = librosa.load(path, sr=SAMPLERATE)
    if len(y) > MAX_LEN: y = y[:MAX_LEN]
    else: y = np.pad(y, (0, MAX_LEN - len(y)))
    return y

def wav_to_mel(wav):
    S = librosa.feature.melspectrogram(y=wav, sr=SAMPLERATE, n_mels=N_MELS)
    return librosa.power_to_db(S, ref=np.max)          # (64, 157)

# =================================================================
# 1.  BUILD (wav_path, label) LIST  →  remap -1/1  →  0/1
# =================================================================
base_dir   = 'data'
csv_files  = glob.glob(os.path.join(base_dir, 'training-*', 'REFERENCE.csv'))
wav_label = []                 # [(wav_abs, 0 or 1), ...]
for csv in csv_files:
    folder = os.path.dirname(csv)
    with open(csv, 'r') as f:
        for line in f:
            line = line.strip()
            if not line: continue
            name, label_str = line.split(',')[0], line.split(',')[1]
            label = int(label_str)          # -1 or 1
            label = 0 if label == -1 else 1 # remap → 0=abnormal, 1=normal
            wav_path = os.path.join(folder, name + '.wav')
            if os.path.isfile(wav_path):
                wav_label.append((wav_path, label))

print('Total WAVs:', len(wav_label))   # should be ~3240

# =================================================================
# 2.  80 / 20 SPLIT
# =================================================================
random.seed(42); random.shuffle(wav_label)
split_at = int(0.8 * len(wav_label))
train_list = wav_label[:split_at]
test_list  = wav_label[split_at:]

def build_set(wav_label_list):
    X, Y = [], []
    for wav_path, label in tqdm(wav_label_list, desc='Building set'):
        mel = wav_to_mel(load_wave(wav_path))
        X.append(mel); Y.append(label)
    return np.array(X)[:, None, :, :], np.array(Y)   # (N,1,64,157)

print('Building train set …'); X_train, y_train = build_set(train_list)
print('Building test set …');  X_test,  y_test  = build_set(test_list)
print(f'Train: {X_train.shape}  Test: {X_test.shape}')

# =================================================================
# 3.  QUICK SAFETY CHECK  (always good to keep)
# =================================================================
print('Unique labels in train:', set(y_train))
print('Unique labels in test :', set(y_test))
assert set(y_train).issubset({0, 1}), 'Labels must be 0 or 1'
assert set(y_test).issubset({0, 1}),  'Labels must be 0 or 1'
```

```python
# ---------- imports ----------
import torch, torch.nn as nn, torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.metrics import accuracy_score
import numpy as np

# ---------- model ----------
class SimpleCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(1, 16, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(16, 32, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(32, 64, 3, padding=1), nn.ReLU(), nn.AdaptiveAvgPool2d((4,4))
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(64*4*4, 32), nn.ReLU(),
            nn.Linear(32, 2)
        )
    def forward(self, x):
        return self.classifier(self.features(x))

# ---------- data ----------
BATCH = 64
train_loader = DataLoader(TensorDataset(torch.tensor(X_train).float(),
                                        torch.tensor(y_train).long()),
                          batch_size=BATCH, shuffle=True)
test_loader  = DataLoader(TensorDataset(torch.tensor(X_test).float(),
                                        torch.tensor(y_test).long()),
                          batch_size=BATCH)

# ---------- train ----------
DEVICE = "cpu"
net = SimpleCNN().to(DEVICE)
crit = nn.CrossEntropyLoss()
opt  = optim.Adam(net.parameters(), lr=1e-3)

for epoch in range(1, 30):          # 10 epochs enough for demo
    net.train()
    for xb, yb in train_loader:
        xb, yb = xb.to(DEVICE), yb.to(DEVICE)
        opt.zero_grad()
        loss = crit(net(xb), yb)
        loss.backward()
        opt.step()

    net.eval()
    with torch.no_grad():
        pred = torch.cat([net(xb).argmax(1) for xb, _ in test_loader])
        true = torch.cat([yb for _, yb in test_loader])
        acc  = accuracy_score(true, pred)
    print(f'Epoch {epoch:02d}  test-acc {acc:.3f}')

torch.save(net.state_dict(), 'baseline.pth')
print('Saved baseline.pth')
```

```python
# ---------- imports ----------
import torch, torch.nn as nn, torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.metrics import accuracy_score
import numpy as np

# ---------- model ----------
import torch
import torch.nn as nn

class DepthwiseSeparableConv(nn.Module):
    def __init__(self, in_ch, out_ch, kernel_size=3, stride=1):
        super().__init__()
        self.dw = nn.Conv2d(in_ch, in_ch, kernel_size, stride, padding=kernel_size//2, groups=in_ch)
        self.pw = nn.Conv2d(in_ch, out_ch, 1)
        self.bn = nn.BatchNorm2d(out_ch)
        self.relu = nn.ReLU()

    def forward(self, x):
        x = self.dw(x)
        x = self.pw(x)
        x = self.bn(x)
        return self.relu(x)

class SEBlock(nn.Module):
    """Squeeze-and-Excitation: channel attention"""
    def __init__(self, ch, reduction=4):
        super().__init__()
        self.squeeze = nn.AdaptiveAvgPool2d(1)
        self.excite = nn.Sequential(
            nn.Linear(ch, ch // reduction),
            nn.ReLU(),
            nn.Linear(ch // reduction, ch),
            nn.Sigmoid()
        )

    def forward(self, x):
        b, c, _, _ = x.shape
        y = self.squeeze(x).view(b, c)
        y = self.excite(y).view(b, c, 1, 1)
        return x * y

class EfficientHeartNet(nn.Module):
    def __init__(self, num_classes=2):
        super().__init__()
        self.features = nn.Sequential(
            # Block 1
            nn.Conv2d(1, 16, 3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(),
            nn.MaxPool2d(2),
            # Block 2: Depthwise Separable
            DepthwiseSeparableConv(16, 32),
            SEBlock(32),
            nn.MaxPool2d(2),
            # Block 3: Standard Conv (more capacity)
            nn.Conv2d(32, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.AdaptiveAvgPool2d((4, 4))
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.3),  # Reduced
            nn.Linear(64*4*4, 32),
            nn.ReLU(),
            nn.Dropout(0.2),  # Reduced
            nn.Linear(32, num_classes)
        )

    def forward(self, x):
        return self.classifier(self.features(x))  # No DropPath

        # Optional: Apply DropPath during training for robustness
        if self.training and self.drop_path_rate > 0:
            keep_prob = 1.0 - self.drop_path_rate
            shape = (x.shape[0],) + (1,) * (x.ndim - 1)
            random_tensor = keep_prob + torch.rand(shape, dtype=x.dtype, device=x.device)
            random_tensor.floor_()  # binarize
            x = x.div(keep_prob) * random_tensor

        x = self.classifier(x)
        return x

# ---------- data ----------
BATCH = 16
train_loader = DataLoader(TensorDataset(torch.tensor(X_train).float(),
                                        torch.tensor(y_train).long()),
                          batch_size=BATCH, shuffle=True)
test_loader  = DataLoader(TensorDataset(torch.tensor(X_test).float(),
                                        torch.tensor(y_test).long()),
                          batch_size=BATCH)

# ---------- train ----------
DEVICE = "cpu"
#net = SimpleCNN().to(DEVICE)
net = EfficientHeartNet().to(DEVICE)
crit = nn.CrossEntropyLoss()
opt  = optim.Adam(net.parameters(), lr=5e-4)

for epoch in range(1, 10):          # 10 epochs enough for demo
    net.train()
    for xb, yb in train_loader:
        xb, yb = xb.to(DEVICE), yb.to(DEVICE)
        opt.zero_grad()
        loss = crit(net(xb), yb)
        loss.backward()
        opt.step()

    net.eval()
    with torch.no_grad():
        pred = torch.cat([net(xb).argmax(1) for xb, _ in test_loader])
        true = torch.cat([yb for _, yb in test_loader])
        acc  = accuracy_score(true, pred)
    print(f'Epoch {epoch:02d}  test-acc {acc:.3f}')

torch.save(net.state_dict(), 'baseline.pth')
print('Saved baseline.pth')
```

```python
# =================================================================
# ADVERSARIAL EVALUATION ON ENTIRE TEST SET — FULL VISUALIZATIONS
# Based on TENFED-TWIN (paper2.pdf): ε=0.05, mel-dB, FGSM
# =================================================================
import os, glob, random, numpy as np, librosa, torch
from tqdm import tqdm
import matplotlib.pyplot as plt
import torch.nn.functional as F
import seaborn as sns
from sklearn.metrics import confusion_matrix, roc_curve, auc
sns.set_style("whitegrid")

# ----------------------------
# Reuse your existing setup
# ----------------------------
SAMPLERATE = 8000; DURATION = 2.0; MAX_LEN = int(DURATION * SAMPLERATE); N_MELS = 64

def load_wave(path):
    y, sr = librosa.load(path, sr=SAMPLERATE)
    if len(y) > MAX_LEN: y = y[:MAX_LEN]
    else: y = np.pad(y, (0, MAX_LEN - len(y)))
    return y

def wav_to_mel(wav):
    S = librosa.feature.melspectrogram(y=wav, sr=SAMPLERATE, n_mels=N_MELS)
    return librosa.power_to_db(S, ref=np.max)  # (64, 157)

# Load test set (you already built test_loader earlier)
# We assume: net, test_loader are defined and model is trained

# ----------------------------
# Load trained model
# ----------------------------
net.load_state_dict(torch.load('baseline.pth', map_location='cpu'))
net.eval()
device = 'cpu'
eps = 0.05

# ----------------------------
# Run FGSM on full test set
# ----------------------------
results = {
    'true_labels': [], 'clean_preds': [], 'adv_preds': [],
    'clean_confs': [], 'adv_confs': [], 'l2_norms': [], 'linf_norms': [],
    'is_successful_attack': []
}
success_examples = []
failure_examples = []
MAX_VIS = 6

print("Running FGSM attack on entire test set...")
for xb, yb in tqdm(test_loader, desc="FGSM Attack"):
    xb, yb = xb.to(device), yb.to(device)
    B = xb.shape[0]

    # --- Clean inference (no grad) ---
    with torch.no_grad():
        out_clean = net(xb)
        prob_clean = F.softmax(out_clean, dim=1)
        pred_clean = out_clean.argmax(dim=1)
        conf_clean = prob_clean.max(dim=1).values

    # --- FGSM ATTACK (MUST be OUTSIDE torch.no_grad!) ---
    xb_adv = xb.clone().requires_grad_(True)      # ✅ Correct: no .detach()
    out = net(xb_adv)
    loss = F.cross_entropy(out, yb)
    loss.backward()                               # ✅ Now works!

    noise = eps * xb_adv.grad.sign()
    xb_adv = torch.clamp(xb_adv + noise, -80, 0).detach()

    # --- Adversarial inference (no grad) ---
    with torch.no_grad():
        out_adv = net(xb_adv)
        prob_adv = F.softmax(out_adv, dim=1)
        pred_adv = out_adv.argmax(dim=1)
        conf_adv = prob_adv.max(dim=1).values

    # Perturbation
    pert = (xb_adv - xb).abs()
    l2_norms = torch.norm(pert.view(B, -1), p=2, dim=1)
    linf_norms = pert.view(B, -1).max(dim=1).values

    # Record results
    for i in range(B):
        true = yb[i].item()
        clean_p = pred_clean[i].item()
        adv_p = pred_adv[i].item()
        clean_c = conf_clean[i].item()
        adv_c = conf_adv[i].item()
        l2 = l2_norms[i].item()
        linf = linf_norms[i].item()

        results['true_labels'].append(true)
        results['clean_preds'].append(clean_p)
        results['adv_preds'].append(adv_p)
        results['clean_confs'].append(clean_c)
        results['adv_confs'].append(adv_c)
        results['l2_norms'].append(l2)
        results['linf_norms'].append(linf)

        clean_correct = (clean_p == true)
        adv_wrong = (adv_p != true)
        success = clean_correct and adv_wrong
        results['is_successful_attack'].append(success)

        if success and len(success_examples) < MAX_VIS:
            success_examples.append((xb[i].cpu().numpy(), xb_adv[i].cpu().numpy(),
                                     pert[i].cpu().numpy(), true, clean_p, adv_p, clean_c, adv_c))
        elif not success and len(failure_examples) < MAX_VIS:
            failure_examples.append((xb[i].cpu().numpy(), xb_adv[i].cpu().numpy(),
                                     pert[i].cpu().numpy(), true, clean_p, adv_p, clean_c, adv_c))

# Convert to numpy
for k in results:
    results[k] = np.array(results[k])

# Metrics
clean_acc = np.mean(results['clean_preds'] == results['true_labels'])
adv_acc = np.mean(results['adv_preds'] == results['true_labels'])
attack_success_rate = np.mean(results['is_successful_attack'])

print(f"\n=== FGSM (ε={eps}) RESULTS ===")
print(f"Clean Accuracy      : {clean_acc:.4f} ({clean_acc*100:.1f}%)")
print(f"Adversarial Acc     : {adv_acc:.4f} ({adv_acc*100:.1f}%)")
print(f"Attack Success Rate : {attack_success_rate:.4f} ({attack_success_rate*100:.1f}%)")
print(f"Average L2 norm     : {np.mean(results['l2_norms']):.4f}")
print(f"Average L∞ norm     : {np.mean(results['linf_norms']):.4f}")

# Paper comparison
print(f"\nPaper (TENFED-TWIN): Clean=0.884, Adv=0.731, Drop=0.153")
print(f"Your Model         : Clean={clean_acc:.3f}, Adv={adv_acc:.3f}, Drop={clean_acc - adv_acc:.3f}")

# =================================================================
# VISUALIZATIONS
# =================================================================

# 1. Example Spectrograms
def plot_examples(examples, title):
    if not examples:
        print(f"No {title.lower()} to show.")
        return
    n = len(examples)
    cols = min(3, n)
    rows = (n + cols - 1) // cols
    fig, axes = plt.subplots(rows, cols * 3, figsize=(15, 3 * rows))
    if rows == 1:
        axes = axes.reshape(1, -1)
    for idx, (x_clean, x_adv, pert, true, clean_p, adv_p, conf_clean, conf_adv) in enumerate(examples):
        r = idx // cols
        c = (idx % cols) * 3
        # Clean
        axes[r, c].imshow(x_clean[0], aspect='auto', origin='lower', cmap='viridis')
        axes[r, c].set_title(f'Clean\nTrue={true}, Pred={clean_p}\nConf={conf_clean:.3f}')
        axes[r, c].axis('off')
        # Adversarial
        axes[r, c+1].imshow(x_adv[0], aspect='auto', origin='lower', cmap='viridis')
        axes[r, c+1].set_title(f'Adv\nPred={adv_p}\nConf={conf_adv:.3f}')
        axes[r, c+1].axis('off')
        # Perturbation
        axes[r, c+2].imshow(pert[0], aspect='auto', origin='lower', cmap='coolwarm', vmin=0, vmax=eps)
        axes[r, c+2].set_title(f'|Δ| (max={pert[0].max():.3f})')
        axes[r, c+2].axis('off')
    plt.suptitle(title, fontsize=16, fontweight='bold')
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.savefig(f'adv_{title.replace(" ", "_").lower()}.png', dpi=200)
    plt.show()

plot_examples(success_examples, "Successful Attacks")
plot_examples(failure_examples, "Failed Attacks")

# 2. Confidence & Accuracy Summary
plt.figure(figsize=(14, 4))
plt.subplot(1, 3, 1)
plt.hist(results['clean_confs'], bins=30, alpha=0.7, label='Clean', color='skyblue')
plt.hist(results['adv_confs'], bins=30, alpha=0.7, label='Adversarial', color='salmon')
plt.xlabel('Confidence'); plt.ylabel('Count'); plt.legend(); plt.title('Confidence Distribution')

plt.subplot(1, 3, 2)
plt.bar(['Clean', 'Adversarial'], [clean_acc, adv_acc], color=['skyblue', 'salmon'])
plt.ylabel('Accuracy'); plt.ylim(0, 1); plt.title('Accuracy Drop')

plt.subplot(1, 3, 3)
plt.hist(results['l2_norms'], bins=30, color='green', alpha=0.7)
plt.axvline(np.mean(results['l2_norms']), color='darkgreen', linestyle='--')
plt.xlabel('L2 Norm'); plt.title('Perturbation L2 Distribution')

plt.tight_layout()
plt.savefig('adv_summary.png', dpi=200)
plt.show()

# 3. Confusion Matrices
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
cm_clean = confusion_matrix(results['true_labels'], results['clean_preds'])
sns.heatmap(cm_clean, annot=True, fmt='d', cmap='Blues')
plt.title(f'Clean Confusion Matrix\n(Acc: {clean_acc:.1%})')

plt.subplot(1, 2, 2)
cm_adv = confusion_matrix(results['true_labels'], results['adv_preds'])
sns.heatmap(cm_adv, annot=True, fmt='d', cmap='Reds')
plt.title(f'Adversarial Confusion Matrix\n(Acc: {adv_acc:.1%})')

plt.tight_layout()
plt.savefig('adv_confusion.png', dpi=200)
plt.show()

# 4. Per-Class Success
unique_labels = np.unique(results['true_labels'])
for lbl in unique_labels:
    mask = results['true_labels'] == lbl
    clean_acc_lbl = np.mean(results['clean_preds'][mask] == lbl)
    adv_acc_lbl = np.mean(results['adv_preds'][mask] == lbl)
    success_lbl = np.mean((results['clean_preds'][mask] == lbl) & (results['adv_preds'][mask] != lbl))
    print(f"Class {lbl}: Clean={clean_acc_lbl:.3f}, Adv={adv_acc_lbl:.3f}, Success={success_lbl:.3f}")
```

```python
# =================================================================
# 9. BOOTSTRAP 95% CI + EDGE METRICS + FLOPs + TIME COMPLEXITY
# =================================================================
from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.utils import resample
import time, os
import numpy as np
import torch

# ---------- Helper: Manual FLOPs Estimation ----------
def count_conv2d_flops(in_ch, out_ch, kernel_size, out_h, out_w):
    """FLOPs for Conv2D = 2 * K_h * K_w * C_in * C_out * H_out * W_out"""
    k_h, k_w = kernel_size if isinstance(kernel_size, tuple) else (kernel_size, kernel_size)
    return 2 * k_h * k_w * in_ch * out_ch * out_h * out_w

def estimate_model_flops(model, input_shape=(1, 1, 64, 157)):
    """Estimate total FLOPs by hooking Conv2D layers."""
    flops = 0
    hooks = []

    def hook_fn(module, input, output):
        nonlocal flops
        if isinstance(module, torch.nn.Conv2d):
            _, _, h_out, w_out = output.shape
            layer_flops = count_conv2d_flops(
                in_ch=module.in_channels,
                out_ch=module.out_channels,
                kernel_size=module.kernel_size,
                out_h=h_out,
                out_w=w_out
            )
            flops += layer_flops

    # Register forward hooks on all Conv2d layers
    for module in model.modules():
        if isinstance(module, torch.nn.Conv2d):
            hooks.append(module.register_forward_hook(hook_fn))

    # Run dummy forward pass
    dummy_input = torch.zeros(input_shape)
    model.eval()
    with torch.no_grad():
        _ = model(dummy_input)

    # Clean up hooks
    for h in hooks:
        h.remove()

    return flops

# ---------- 1. 95% Bootstrap CI for Accuracy & AUC ----------
def bootstrap_ci(y_true, y_pred, metric_func, n_bootstrap=2000, seed=42):
    np.random.seed(seed)
    n = len(y_true)
    bootstrap_stats = []
    for i in range(n_bootstrap):
        idx = resample(range(n), n_samples=n, random_state=seed + i)
        bootstrap_stats.append(metric_func(y_true[idx], y_pred[idx]))
    lower = np.percentile(bootstrap_stats, 2.5)
    upper = np.percentile(bootstrap_stats, 97.5)
    point = metric_func(y_true, y_pred)
    return lower, point, upper

# ---------- Evaluate on TEST set ----------
net.eval()
all_pred, all_true, all_probs = [], [], []
with torch.no_grad():
    for xb, yb in test_loader:
        logits = net(xb)
        probs = torch.softmax(logits, dim=1)[:, 1]   # P(class=1)
        all_pred.append(logits.argmax(1).cpu().numpy())
        all_true.append(yb.cpu().numpy())
        all_probs.append(probs.cpu().numpy())

true = np.concatenate(all_true)
pred = np.concatenate(all_pred)
probs = np.concatenate(all_probs)

# Compute CIs
lower_acc, point_acc, upper_acc = bootstrap_ci(true, pred, accuracy_score)
lower_auc, point_auc, upper_auc = bootstrap_ci(true, probs, roc_auc_score)

print("Bootstrap 95 % CI (2000 re-samples):")
print(f"Accuracy : {point_acc:.3f}  [95 % CI: {lower_acc:.3f} – {upper_acc:.3f}]")
print(f"AUC      : {point_auc:.3f}  [95 % CI: {lower_auc:.3f} – {upper_auc:.3f}]")

# ---------- 2. EDGE METRICS ----------
# RAM (model parameters only, 32-bit floats)
param_count = sum(p.numel() for p in net.parameters())
ram_model = param_count * 4 / 1024  # bytes → kB
print(f"RAM (model)     : {ram_model:.1f} kB")

# Flash (saved state dict size)
torch.save(net.state_dict(), '/tmp/model.pth')
flash_model = os.path.getsize('/tmp/model.pth') / 1024  # bytes → kB
print(f"Flash (model)   : {flash_model:.1f} kB")

# FLOPs & Time Complexity
flops = estimate_model_flops(net, input_shape=(1, 1, 64, 157))
print(f"FLOPs           : {flops:.0f} ({flops/1e6:.1f} MFLOPs)")
print("Time Complexity : Linear in FLOPs (O(FLOPs)); dominated by convolutional layers")

# Latency (median over 100 runs on CPU)
net.cpu()
dummy = torch.randn(1, 1, 64, 157)
times = []
for _ in range(100):
    start = time.perf_counter()
    with torch.no_grad():
        _ = net(dummy)
    times.append((time.perf_counter() - start) * 1000)  # ms
latency = np.median(times)
print(f"Latency (CPU)   : {latency:.2f} ms  (median, n=100)")
```

```python
# ---------- Bayesian Optimisation for α (Supplementary Algo 1) ----------
import numpy as np

# ---------- 1. GP surrogate (Matern kernel) ----------
def matern_kernel(x1, x2, l=0.3, nu=2.5):
    d = np.abs(x1 - x2) / l
    if nu == 2.5:
        return (1 + np.sqrt(5)*d + 5/3*d**2) * np.exp(-np.sqrt(5)*d)
    else:
        return np.exp(-0.5 * d**2)  # RBF fallback

def expected_improvement(X, y, xi=0.01):
    μ = np.mean(y)
    σ = np.std(y) if np.std(y) > 0 else 1e-6
    best = np.min(y)
    ei = np.maximum(0, best - μ - xi) / σ
    return ei

# ---------- 2. Objective: negative SNR (minimise) ----------
def snr_proxy(alpha):
    t = 16.0                               # days
    sigma = 50e-6 * (t ** -alpha)         # power-law decay
    snr = 36 + 10 * np.log10(sigma / 50e-6)  # dB proxy
    return -snr                            # negative → minimise

# ---------- 3. Bayesian loop (no external libs) ----------
bounds = (0.1, 2.0)                       # prior range for α
n_calls = 30                              # budget
X_obs = np.linspace(bounds[0], bounds[1], 5)  # initial grid
y_obs = np.array([snr_proxy(x) for x in X_obs])

for i in range(n_calls - len(X_obs)):
    K = matern_kernel(X_obs[:, None], X_obs[None, :])
    L = np.linalg.cholesky(K + 1e-6 * np.eye(len(X_obs)))
    alpha_next = X_obs[np.argmax(expected_improvement(X_obs, y_obs))]
    y_next = snr_proxy(alpha_next)
    X_obs = np.append(X_obs, alpha_next)
    y_obs = np.append(y_obs, y_next)

alpha_opt = X_obs[np.argmin(y_obs)]
print('Optimal α:', alpha_opt, 'SNR at day 16:', -np.min(y_obs), 'dB')
```

```python
# =================================================================
# EDGE DEPLOYMENT SIMULATOR — USING YOUR EXACT MODEL DEFINITION
# =================================================================
import os, time, numpy as np, torch
from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.utils import resample

# ---------- YOUR EXACT MODEL ----------
class SimpleCNN(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.features = torch.nn.Sequential(
            torch.nn.Conv2d(1, 16, 3, padding=1),
            torch.nn.ReLU(),
            torch.nn.MaxPool2d(2),
            torch.nn.Conv2d(16, 32, 3, padding=1),
            torch.nn.ReLU(),
            torch.nn.MaxPool2d(2),
            torch.nn.Conv2d(32, 64, 3, padding=1),
            torch.nn.ReLU(),
            torch.nn.AdaptiveAvgPool2d((4, 4))  # ← Critical: (4,4), not (1,1)!
        )
        self.classifier = torch.nn.Sequential(
            torch.nn.Flatten(),
            torch.nn.Linear(64 * 4 * 4, 32),  # ← 64*4*4 = 1024
            torch.nn.ReLU(),
            torch.nn.Linear(32, 2)
        )
    def forward(self, x):
        return self.classifier(self.features(x))

# ---------- FLOPs ESTIMATION ----------
def count_conv2d_flops(in_ch, out_ch, kernel_size, out_h, out_w):
    k_h, k_w = kernel_size if isinstance(kernel_size, tuple) else (kernel_size, kernel_size)
    return 2 * k_h * k_w * in_ch * out_ch * out_h * out_w

def estimate_flops(model, input_shape=(1, 1, 64, 157)):
    flops = 0
    hooks = []
    def hook_fn(module, input, output):
        nonlocal flops
        if isinstance(module, torch.nn.Conv2d):
            _, _, h, w = output.shape
            flops += count_conv2d_flops(
                module.in_channels,
                module.out_channels,
                module.kernel_size,
                h, w
            )
    for m in model.modules():
        if isinstance(m, torch.nn.Conv2d):
            hooks.append(m.register_forward_hook(hook_fn))
    dummy = torch.zeros(input_shape)
    model.eval()
    with torch.no_grad():
        _ = model(dummy)
    for h in hooks: h.remove()
    return flops

# ---------- BOOTSTRAP CI ----------
def bootstrap_ci(y_true, y_pred, metric_func, n_bootstrap=2000, seed=42):
    np.random.seed(seed)
    n = len(y_true)
    stats = []
    for i in range(n_bootstrap):
        idx = resample(range(n), n_samples=n, random_state=seed + i)
        stats.append(metric_func(y_true[idx], y_pred[idx]))
    lower = np.percentile(stats, 2.5)
    upper = np.percentile(stats, 97.5)
    point = metric_func(y_true, y_pred)
    return lower, point, upper

# ---------- EDGE EVALUATION ----------
def evaluate_on_edge(model_path, test_loader):
    print(f"\n{'='*60}")
    print(f"DEPLOYING: {model_path}")
    print(f"{'='*60}")

    # Load model
    net = SimpleCNN()
    net.load_state_dict(torch.load(model_path, map_location='cpu'))
    net.eval()
    net.cpu()

    # Evaluate
    all_true, all_pred, all_probs = [], [], []
    with torch.no_grad():
        for xb, yb in test_loader:
            xb, yb = xb.cpu(), yb.cpu()
            logits = net(xb)
            probs = torch.softmax(logits, dim=1)[:, 1]
            all_true.append(yb.numpy())
            all_pred.append(logits.argmax(1).numpy())
            all_probs.append(probs.numpy())

    y_true = np.concatenate(all_true)
    y_pred = np.concatenate(all_pred)
    y_prob = np.concatenate(all_probs)

    acc_low, acc_pt, acc_up = bootstrap_ci(y_true, y_pred, accuracy_score)
    auc_low, auc_pt, auc_up = bootstrap_ci(y_true, y_prob, roc_auc_score)
    print(f"Accuracy : {acc_pt:.3f}  [95% CI: {acc_low:.3f} – {acc_up:.3f}]")
    print(f"AUC      : {auc_pt:.3f}  [95% CI: {auc_low:.3f} – {auc_up:.3f}]")

    # Model size
    param_bytes = sum(p.numel() * 4 for p in net.parameters())
    ram_kb = param_bytes / 1024
    torch.save(net.state_dict(), '/content/tmp_model.pth')
    flash_kb = os.path.getsize('/content/tmp_model.pth') / 1024
    print(f"RAM (model)     : {ram_kb:.1f} kB")
    print(f"Flash (model)   : {flash_kb:.1f} kB")

    # FLOPs & Latency
    flops = estimate_flops(net, input_shape=(1, 1, 64, 157))
    print(f"FLOPs           : {flops:.0f} ({flops/1e6:.1f} MFLOPs)")

    dummy = torch.randn(1, 1, 64, 157)
    times_ms = []
    for _ in range(100):
        start = time.perf_counter()
        with torch.no_grad():
            _ = net(dummy)
        times_ms.append((time.perf_counter() - start) * 1000)
    latency_ms = np.median(times_ms)
    print(f"Latency (CPU)   : {latency_ms:.2f} ms  (median, n=100)")
    print(f"Throughput      : {1000/latency_ms:.1f} inferences/sec")

# ---------- RUN EVALUATION ----------
# Ensure test_loader is defined (from your data loading code)

evaluate_on_edge('/content/baseline.pth', test_loader)
evaluate_on_edge('/content/federated.pth', test_loader)
```

```python
# =================================================================
# EDGE DEPLOYMENT SIMULATOR — FOR EFFICIENT-HEARTNET
# =================================================================
import os, time, numpy as np, torch
from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.utils import resample

# ---------- YOUR OPTIMIZED MODEL ----------
class DepthwiseSeparableConv(torch.nn.Module):
    def __init__(self, in_ch, out_ch, kernel_size=3, stride=1):
        super().__init__()
        self.dw = torch.nn.Conv2d(in_ch, in_ch, kernel_size, stride, padding=kernel_size//2, groups=in_ch)
        self.pw = torch.nn.Conv2d(in_ch, out_ch, 1)
        self.bn = torch.nn.BatchNorm2d(out_ch)
        self.relu = torch.nn.ReLU()
    def forward(self, x):
        x = self.dw(x)
        x = self.pw(x)
        x = self.bn(x)
        return self.relu(x)

class SEBlock(torch.nn.Module):
    def __init__(self, ch, reduction=4):
        super().__init__()
        self.squeeze = torch.nn.AdaptiveAvgPool2d(1)
        self.excite = torch.nn.Sequential(
            torch.nn.Linear(ch, ch // reduction),
            torch.nn.ReLU(),
            torch.nn.Linear(ch // reduction, ch),
            torch.nn.Sigmoid()
        )
    def forward(self, x):
        b, c, _, _ = x.shape
        y = self.squeeze(x).view(b, c)
        y = self.excite(y).view(b, c, 1, 1)
        return x * y

class EfficientHeartNet(torch.nn.Module):
    def __init__(self, num_classes=2, drop_path_rate=0.1):
        super().__init__()
        self.features = torch.nn.Sequential(
            # Block 1: Standard conv
            torch.nn.Conv2d(1, 16, 3, padding=1),
            torch.nn.BatchNorm2d(16),
            torch.nn.ReLU(),
            torch.nn.MaxPool2d(2),
            # Block 2: DepthwiseSeparable + SE
            DepthwiseSeparableConv(16, 32),
            SEBlock(32),
            torch.nn.MaxPool2d(2),
            # Block 3: DepthwiseSeparable + SE
            DepthwiseSeparableConv(32, 64),
            SEBlock(64),
            torch.nn.AdaptiveAvgPool2d((4, 4))
        )
        self.classifier = torch.nn.Sequential(
            torch.nn.Flatten(),
            torch.nn.Dropout(0.5),
            torch.nn.Linear(64 * 4 * 4, 32),
            torch.nn.ReLU(),
            torch.nn.Dropout(0.3),
            torch.nn.Linear(32, num_classes)
        )
        self.drop_path_rate = drop_path_rate

    def forward(self, x):
        x = self.features(x)
        if self.training and self.drop_path_rate > 0:
            keep_prob = 1.0 - self.drop_path_rate
            shape = (x.shape[0],) + (1,) * (x.ndim - 1)
            random_tensor = keep_prob + torch.rand(shape, dtype=x.dtype, device=x.device)
            random_tensor.floor_()
            x = x.div(keep_prob) * random_tensor
        x = self.classifier(x)
        return x

# ---------- FLOPs ESTIMATION (handles DepthwiseSeparableConv) ----------
def estimate_flops(model, input_shape=(1, 1, 64, 157)):
    flops = 0
    hooks = []

    def hook_fn(module, input, output):
        nonlocal flops
        if isinstance(module, torch.nn.Conv2d):
            _, _, h_out, w_out = output.shape
            k_h, k_w = module.kernel_size
            # Standard Conv: C_in * C_out * K^2 * H * W
            if module.groups == 1:
                flops += 2 * k_h * k_w * module.in_channels * module.out_channels * h_out * w_out
            # Depthwise Conv: C_in * K^2 * H * W (since C_in = groups)
            elif module.groups == module.in_channels:
                flops += 2 * k_h * k_w * module.in_channels * h_out * w_out
            # Pointwise Conv: C_in * C_out * H * W (1x1)
            elif k_h == 1 and k_w == 1:
                flops += 2 * module.in_channels * module.out_channels * h_out * w_out

    for m in model.modules():
        if isinstance(m, torch.nn.Conv2d):
            hooks.append(m.register_forward_hook(hook_fn))

    dummy = torch.zeros(input_shape)
    model.eval()
    with torch.no_grad():
        _ = model(dummy)
    for h in hooks:
        h.remove()
    return flops

# ---------- BOOTSTRAP CI ----------
def bootstrap_ci(y_true, y_pred, metric_func, n_bootstrap=2000, seed=42):
    np.random.seed(seed)
    n = len(y_true)
    stats = []
    for i in range(n_bootstrap):
        idx = resample(range(n), n_samples=n, random_state=seed + i)
        stats.append(metric_func(y_true[idx], y_pred[idx]))
    lower = np.percentile(stats, 2.5)
    upper = np.percentile(stats, 97.5)
    point = metric_func(y_true, y_pred)
    return lower, point, upper

# ---------- EDGE EVALUATION ----------
def evaluate_on_edge(model_path, test_loader):
    print(f"\n{'='*60}")
    print(f"DEPLOYING: {model_path}")
    print(f"{'='*60}")

    net = EfficientHeartNet().to(DEVICE)
    net.load_state_dict(torch.load(model_path, map_location='cpu'))
    net.eval()
    net.cpu()

    all_true, all_pred, all_probs = [], [], []
    with torch.no_grad():
        for xb, yb in test_loader:
            xb, yb = xb.cpu(), yb.cpu()
            logits = net(xb)
            probs = torch.softmax(logits, dim=1)[:, 1]
            all_true.append(yb.numpy())
            all_pred.append(logits.argmax(1).numpy())
            all_probs.append(probs.numpy())

    y_true = np.concatenate(all_true)
    y_pred = np.concatenate(all_pred)
    y_prob = np.concatenate(all_probs)

    acc_low, acc_pt, acc_up = bootstrap_ci(y_true, y_pred, accuracy_score)
    auc_low, auc_pt, auc_up = bootstrap_ci(y_true, y_prob, roc_auc_score)
    print(f"Accuracy : {acc_pt:.3f}  [95% CI: {acc_low:.3f} – {acc_up:.3f}]")
    print(f"AUC      : {auc_pt:.3f}  [95% CI: {auc_low:.3f} – {auc_up:.3f}]")

    param_bytes = sum(p.numel() * 4 for p in net.parameters())
    ram_kb = param_bytes / 1024
    torch.save(net.state_dict(), '/content/tmp_model.pth')
    flash_kb = os.path.getsize('/content/tmp_model.pth') / 1024
    print(f"RAM (model)     : {ram_kb:.1f} kB")
    print(f"Flash (model)   : {flash_kb:.1f} kB")

    flops = estimate_flops(net, input_shape=(1, 1, 64, 157))
    print(f"FLOPs           : {flops:.0f} ({flops/1e6:.1f} MFLOPs)")

    dummy = torch.randn(1, 1, 64, 157)
    times_ms = []
    for _ in range(100):
        start = time.perf_counter()
        with torch.no_grad():
            _ = net(dummy)
        times_ms.append((time.perf_counter() - start) * 1000)
    latency_ms = np.median(times_ms)
    print(f"Latency (CPU)   : {latency_ms:.2f} ms  (median, n=100)")
    print(f"Throughput      : {1000/latency_ms:.1f} inferences/sec")

# ---------- RUN ----------
# Make sure test_loader is defined

evaluate_on_edge('/content/baseline.pth', test_loader)              # optimized baseline
evaluate_on_edge('/content/federated.pth', test_loader)   # optimized FL
```

```python
# ================================================================
# 6. TENG CONSTITUTIVE MODEL
# ================================================================
class TENG:
    def __init__(self, S=4e-4, sigma=50e-6, d0=25e-6, C0=10e-12):
        self.S, self.sigma, self.d0, self.C0 = S, sigma, d0, C0
    def voltage(self, D):
        num = self.S * self.sigma
        den = (self.d0 + D) / (8.85e-12 * self.S) + self.C0
        return num / den

teng = TENG()
D = np.linspace(0, 0.5e-3, 200)  # 0-0.5 mm
plt.figure()
plt.plot(D*1e3, teng.voltage(D)*1e3)
plt.xlabel('Displacement (mm)'); plt.ylabel('Vout (mV)')
plt.title('TENG fast saturation'); plt.grid()
plt.savefig('teng_curve.png', dpi=300)
plt.show()
```

```python
# ================================================================
# 7. FEDERATED LEARNING DEMO (2 clients)
# ================================================================
import copy
CLIENTS = 2
ROUNDS  = 5
EPOCHS_PER_ROUND = 3

# split training data
from torch.utils.data import random_split
subsets = random_split(list(train_loader.dataset), [0.5, 0.5])
client_loaders = [DataLoader(s, BATCH, shuffle=True) for s in subsets]

global_net = SimpleCNN().to(DEVICE)
client_nets = [copy.deepcopy(global_net) for _ in range(CLIENTS)]

for rnd in range(ROUNDS):
    print(f'--- Round {rnd+1} ---')
    # local training
    for c in range(CLIENTS):
        net = client_nets[c]
        opt = torch.optim.Adam(net.parameters(), lr=LR)
        net.train()
        for epoch in range(EPOCHS_PER_ROUND):
            for xb, yb in client_loaders[c]:
                xb, yb = xb.to(DEVICE), yb.to(DEVICE)
                opt.zero_grad()
                loss = crit(net(xb), yb)
                loss.backward(); opt.step()
    # federated averaging
    with torch.no_grad():
        global_dict = global_net.state_dict()
        for key in global_dict:
            global_dict[key] = torch.stack([client_nets[c].state_dict()[key] for c in range(CLIENTS)], 0).mean(0)
        global_net.load_state_dict(global_dict)
        for c in range(CLIENTS):
            client_nets[c].load_state_dict(global_dict)
    # quick eval
    global_net.eval()
    all_pred, all_true = [], []
    with torch.no_grad():
        for xb, yb in test_loader:
            xb, yb = xb.to(DEVICE), yb.to(DEVICE)
            all_pred.append(global_net(xb).argmax(1).cpu())
            all_true.append(yb.cpu())
    all_pred, all_true = torch.cat(all_pred), torch.cat(all_true)
    acc = accuracy_score(all_true, all_pred)
    print(f'Round {rnd+1}  global accuracy {acc:.3f}')

torch.save(global_net.state_dict(), 'federated.pth')
```
